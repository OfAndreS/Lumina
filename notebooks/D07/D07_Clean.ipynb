#### **01 ) - Tratamento dos dados**

- 

---

```python
df_d07_clean = df_d07.copy()
```

#### 2.1 - *Dropando colunas*

```python

# Dropando a coluna 'dataset_split'. Uma vez que ela representa o resultado de um train_test_split indesejado.
df_d07_clean = df_d07_clean.drop(columns=['dataset_split'])
# Dropando a coluna 'education_num'. Uma vez que ela representa um identificador único para a coluna '4_education' criando redundância nos dados. 
df_d07_clean = df_d07_clean.drop(columns=['education_num'])
```

#### 2.2 - *Renomeando colunas*

```python

# Melhorando a legibilidade da coluna 'fnlwgt'
df_d07_clean = df_d07_clean.rename(columns={'fnlwgt' : 'final_weight'})
# Renomeando a coluna alvo para TARGET
df_d07_clean = df_d07_clean.rename(columns={"income":'TARGET'})
# Aplicando a contagem padrão nas colunas
df_d07_clean.columns = [f"{idx}_{col}" for idx, col in enumerate(df_d07_clean.columns, start=1)]
```

#### 2.3 - *Tratamento para valores ausentes*

```python

# O dataset Adult marca os seus valores ausentes com o símbulo '?'. Portanto, optei por melhorar a legibilidade trocando o valor para 'Missing_Value'.
df_d07_clean = df_d07_clean.apply(lambda col: col.str.strip() if col.dtype == 'object' else col)
mask_1 = {'?': 'Missing_Value'}
df_d07_clean = df_d07_clean.replace(mask_1)
```

#### 2.4 - *Tratamento para valores duplicados*

```python
# Como existem apenas 101 linhas duplicadas em um conjunto de dados com mais de 48 mil registros, optei por removê-las, mantendo apenas a primeira instância.
print(f"| Quantitade de linhas duplicadas: {df_d07_clean[df_d07_clean.duplicated(keep=False)].sort_values(by=list(df_d07_clean.columns)).shape[0]}")
```

```python
df_d07_clean = df_d07_clean.drop_duplicates(keep='first')
```

#### 2.5 - *Tratamento para colunas específicas*

```python

# Modifica o valor da variável alvo.
mask_2 = {'<=50K': 0, '>50K': 1}
df_d07_clean['14_TARGET'] = df_d07_clean['14_TARGET'].map(mask_2)
df_d07_clean['14_TARGET'].unique()
```

#### 2.6 - *Salvando o resultado*

```python

df_d07_clean.to_csv("data/input/D07_F1.csv", index=False)
df_d07_clean.head()
```
