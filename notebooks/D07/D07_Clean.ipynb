```python
from sklearn.preprocessing import OneHotEncoder
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# D07 - Adult
df_d07 = pd.read_csv("raw/Adult_dataset.csv")
df_d07.head()
```

---

#### **01 ) - Tratamento dos dados**

- 

---

```python

df_d07_clean = df_d07.copy()
df_d07_clean = df_d07_clean.drop(columns=['dataset_split'])
df_d07_clean = df_d07_clean.rename(columns={'fnlwgt' : 'final_weight'})
df_d07_clean.columns = [f"{idx}_{col}" for idx, col in enumerate(df_d07_clean.columns, start=1)]
```

```python
# Substitui todos os valores ausentes presentes nas variáveis categóricas "2_workclass" e "7_occupation" por Missing_Value.
df_d07_clean = df_d07_clean.apply(lambda col: col.str.strip() if col.dtype == 'object' else col)
mask_1 = {'?': 'Missing_Value'}
df_d07_clean = df_d07_clean.replace(mask_1)
```

```python
# Modifica o valor da variável alvo.
mask_2 = {'<=50K': 0, '>50K': 1}
df_d07_clean = df_d07_clean.replace(mask_2)
df_d07_clean['15_income'].unique()
```

```python
df_d07_clean.to_csv("Input/D07_V1.csv", index=False)
df_d07_clean.head()
```
